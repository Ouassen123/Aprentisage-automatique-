```python
documents = []

for _, row in df.iterrows():

    document = f"""
    ID : {row['id']}
    Catégorie : {row['categorie']}
    Période : {row['periode']}
    Sujet : {row['sujet']}
    Crédits : {row['credits']}
    Responsable : {row['responsable']}
    Contenu : {row['contenu']}
    """

    documents.append(document)

print(documents[0])
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
```

```python
document_embeddings = model.encode(documents)

print("Nombre de documents :", len(document_embeddings))
print("Taille d'un vecteur :", len(document_embeddings[0]))
```

```python
def rechercher(question, top_k=3):

    question_embedding = model.encode([question])

    scores = cosine_similarity(
        question_embedding,
        document_embeddings
    )[0]

    indices = np.argsort(scores)[::-1][:top_k]

    resultats = []

    for i in indices:
        resultats.append({
            "score": scores[i],
            "document": documents[i]
        })

    return resultats
```

```python
question = "Quel professeur enseigne le Machine Learning ?"

resultats = rechercher(question)

for resultat in resultats:

    print("=" * 60)
    print("Score :", round(resultat["score"], 3))
    print(resultat["document"])
```

```python
questions = [
    "Quels sont les modules du semestre 2 ?",
    "Qui enseigne le Deep Learning ?",
    "Combien de crédits possède le module Digital Twin ?",
    "Que se passe-t-il après trois absences non justifiées ?",
    "Quand commencent les cours ?",
    "Quand ont lieu les examens du deuxième semestre ?"
]

for question in questions:

    print("\nQUESTION :", question)

    resultats = rechercher(question, top_k=1)

    print(resultats[0]["document"])
```

```python
def construire_contexte(question, top_k=3):

    resultats = rechercher(question, top_k)

    contexte = ""

    for i, resultat in enumerate(resultats):

        contexte += f"""
DOCUMENT {i+1}

{resultat['document']}

"""

    return contexte
```

```python
question = "Qui enseigne le Digital Twin ?"

contexte = construire_contexte(question)

print(contexte)
```
