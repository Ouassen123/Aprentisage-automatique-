```python
documents = []

for _, row in df.iterrows():

    document = f"""
    ID : {row['id']}
    Catégorie : {row['categorie']}
    Période : {row['periode']}
    Sujet : {row['sujet']}
    Crédits : {row['credits']}
    Responsable : {row['responsable']}
    Contenu : {row['contenu']}
    """

    documents.append(document)

print(documents[0])
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
```

```python
document_embeddings = model.encode(documents)

print("Nombre de documents :", len(document_embeddings))
print("Taille d'un vecteur :", len(document_embeddings[0]))
```

```python
def rechercher(question, top_k=3):

    question_embedding = model.encode([question])

    scores = cosine_similarity(
        question_embedding,
        document_embeddings
    )[0]

    indices = np.argsort(scores)[::-1][:top_k]

    resultats = []

    for i in indices:
        resultats.append({
            "score": scores[i],
            "document": documents[i]
        })

    return resultats
```

```python
question = "Quel professeur enseigne le Machine Learning ?"

resultats = rechercher(question)

for resultat in resultats:

    print("=" * 60)
    print("Score :", round(resultat["score"], 3))
    print(resultat["document"])
```

```python
def rechercher_avec_filtre(question, top_k=3, categorie=None):

    question_embedding = model.encode([question])

    scores = cosine_similarity(
        question_embedding,
        document_embeddings
    )[0]

    indices = np.argsort(scores)[::-1]

    resultats = []

    for i in indices:

        # Filtrage par catégorie
        if categorie is not None:
            if df.iloc[i]["categorie"] != categorie:
                continue

        resultats.append({
            "score": scores[i],
            "document": documents[i],
            "index": i
        })

        if len(resultats) >= top_k:
            break

    return resultats
```

```python
resultats = rechercher_avec_filtre(
    "Quels sont les modules du semestre 2 ?",
    top_k=5,
    categorie="Master Intelligence Artificielle"
)

for resultat in resultats:

    print("=" * 60)
    print("Score :", round(resultat["score"], 3))
    print(resultat["document"])
```

```python
def rechercher_modules_semestre(question, semestre, top_k=5):

    question_embedding = model.encode([question])

    scores = cosine_similarity(
        question_embedding,
        document_embeddings
    )[0]

    indices = np.argsort(scores)[::-1]

    resultats = []

    for i in indices:

        row = df.iloc[i]

        if row["categorie"] != "Master Intelligence Artificielle":
            continue

        if row["periode"] != semestre:
            continue

        resultats.append({
            "score": scores[i],
            "document": documents[i],
            "sujet": row["sujet"],
            "responsable": row["responsable"],
            "credits": row["credits"]
        })

        if len(resultats) >= top_k:
            break

    return resultats
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM, pipeline
import torch

MODEL_NAME = "google/flan-t5-base"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model_llm = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

def generer_reponse(prompt):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=1024
    )

    with torch.no_grad():
        outputs = model_llm.generate(
            **inputs,
            max_new_tokens=150,
            temperature=0.2,
            do_sample=False
        )

    reponse = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return reponse.strip()

print("LLM chargé avec succès !")
```

```python
def rechercher_rag(question, top_k=3, seuil=0.35):

    question_embedding = model.encode([question])

    scores = cosine_similarity(
        question_embedding,
        document_embeddings
    )[0]

    indices = np.argsort(scores)[::-1]

    resultats = []

    for i in indices[:top_k]:

        score = float(scores[i])

        if score >= seuil:

            resultats.append({
                "score": score,
                "document": documents[i],
                "id": df.iloc[i]["id"],
                "sujet": df.iloc[i]["sujet"],
                "categorie": df.iloc[i]["categorie"]
            })

    return resultats
```

```python
seuil=0.35
```

```python
def construire_contexte(question, top_k=3):

    resultats = rechercher_rag(
        question,
        top_k=top_k
    )

    if not resultats:
        return None, []

    contexte = ""

    for i, r in enumerate(resultats):

        contexte += f"""
DOCUMENT {i+1}
ID : {r['id']}
Sujet : {r['sujet']}
Catégorie : {r['categorie']}

{r['document']}

"""

    return contexte, resultats
```

```python
def mon_rag(question):

    question = question.strip()

    if not question:
        return "Veuillez poser une question.", []

    # 1. Recherche
    contexte, resultats = construire_contexte(
        question,
        top_k=3
    )

    # 2. Aucun document suffisamment pertinent
    if contexte is None:

        return (
            "❌ Je ne trouve pas cette information "
            "dans ma base de connaissances.",
            []
        )

    # 3. Construire le prompt
    prompt = f"""
Tu es un assistant universitaire.

Tu dois répondre UNIQUEMENT avec les informations
présentes dans les documents fournis.

NE PAS inventer d'information.

Si la réponse à la question n'est pas présente
dans les documents, réponds exactement :

Je ne trouve pas cette information dans ma base de connaissances.

DOCUMENTS :

{contexte}

QUESTION :

{question}

RÉPONSE :
"""

    # 4. Génération avec FLAN-T5
    resultat_llm = generer_reponse(prompt)

    # 5. Sources
    sources = []

    for r in resultats:

        sources.append(
            f"{r['id']} — {r['sujet']} "
            f"(similarité : {r['score']:.2f})"
        )

    return resultat_llm, sources
```

```python
reponse, sources = mon_rag(
    "Qui enseigne le Machine Learning ?"
)

print("RÉPONSE :")
print(reponse)

print("\nSOURCES :")

for source in sources:
    print("-", source)
```

```python
import gradio as gr
```

```python
def chat_rag(message):

    reponse, sources = mon_rag(message)

    texte_sources = ""

    if sources:

        texte_sources = "\n\n### 📚 Sources utilisées\n"

        for source in sources:
            texte_sources += f"- {source}\n"

    return reponse + texte_sources
```

```python
with gr.Blocks(title="Université - Assistant RAG") as demo:

    gr.Markdown(
        """
        # 🤖 Bonjour, je suis votre assistant RAG

        Je peux répondre à vos questions en utilisant
        les informations contenues dans ma base de connaissances.

        ### 💬 Posez-moi une question

        Exemples :

        - Qui enseigne le Machine Learning ?
        - Qui enseigne le Digital Twin ?
        - Combien de crédits possède le Digital Twin ?
        - Quels sont les modules du semestre 2 ?
        - Quand commencent les cours ?
        - Quand ont lieu les examens S2 ?

        ⚠️ Je ne dois pas inventer une information
        qui n'existe pas dans ma base.
        """
    )

    chatbot = gr.Chatbot(
        label="Assistant RAG",
        height=450
    )

    message = gr.Textbox(
        label="Votre question",
        placeholder="Posez votre question ici..."
    )

    bouton = gr.Button(
        "Envoyer 🚀"
    )

    def repondre(message, historique):

        reponse = chat_rag(message)

        historique = historique or []

        historique.append(
            {"role": "user", "content": message}
        )

        historique.append(
            {"role": "assistant", "content": reponse}
        )

        return historique, ""

    bouton.click(
        repondre,
        inputs=[message, chatbot],
        outputs=[chatbot, message]
    )

    message.submit(
        repondre,
        inputs=[message, chatbot],
        outputs=[chatbot, message]
    )

demo.launch()
```
